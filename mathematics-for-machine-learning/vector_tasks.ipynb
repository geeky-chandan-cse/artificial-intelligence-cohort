### Q2. Find the magnitude of the vector C = [2, -3, 1].

````python
vector_c = np.array([2, -3, 1])
# Formula = sqrt(a^2 + b^2 + c^3)
# Method 1 --> Brute force
magnitude_of_vector_c = np.sqrt(np.sum(np.square(vector_c)))

# Mehod 2 --> using dot product of the same vector
mag_using_dot = np.sqrt(vector_c @ vector_c)

# Method 3 --> using built in numpy method - recommended
mag_using_norm = np.linalg.norm(vector_c)
print(magnitude_of_vector_c, mag_using_dot, mag_using_norm)
````

### Q3. Given the vectors D = [1, 2, 3] and E = [4, 5, 6], calculate the cosine of the angle between them.

````python
vector_d = np.array([1, 2, 3])
vector_e = np.array([4, 5, 5 ])

# Step 1: Calculate the dot product
# dot_product = vector_d @ vector_e
dot_product = np.dot(vector_d, vector_e)

# Step 2: Calculate the magnitudes
mag_d = np.linalg.norm(vector_d)
mag_e = np.linalg.norm(vector_e)

# Step 3: Divide dot product by the multiplied magnitudes
cos_theta = dot_product / (mag_d * mag_e)
print("Cosine of the angle:", cos_theta)

# Advanced steps :- find angle b/w them
# get value in radions using cos inverse in radians
angle_radians = np.arccos(cos_theta)

# Convert radians to degree
angle_degrees = np.degrees(angle_radians)
print(f"Angle in radians: {angle_radians:.4f}")
print(f"Angle in degrees: {angle_degrees:.2f}°")
````

### Q6. Given the vectors I = [1, 2, 3] and J = [4, 5, 6], calculate the projection of vector I onto vector J.

````python
vector_i = np.array([1, 2, 3])
vector_j = np.array([4, 5, 5 ])

# Calculate projection using the simplified dot product approach
projection = (np.dot(vector_i, vector_j) / np.dot(vector_j, vector_j)) * vector_j

print(np.round(projection, 4))
````

### Q7. Consider the 2D - position vector a = (1, 3). If b is the point (5, 4), find the vector `ab` and the vector sum `a + ab`. Provide a sketch plot.
Note: Search for quiver, to plot an arrow.

````python
import numpy as np
import matplotlib.pyplot as plt

# Define vectors
a = np.array([1, 3])
b = np.array([5, 4])
ab = b - a # traveling FROM point b TO point a
sum_vec = a + ab

# Setup the coordinate grid
plt.figure(figsize=(7, 7))
plt.grid( linestyle='dotted', alpha=0.6)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='Vector a')
plt.quiver(a[0], a[1], ab[0], ab[1], angles='xy', scale_units='xy', scale=1, color='orange', label='Vector ab')
plt.quiver(0, 0, sum_vec[0], sum_vec[1], angles='xy', scale_units='xy', scale=1, color='green', label='Sum a + ab')

# Add point markers for reference
plt.autoscale(False)
plt.plot(a[0], a[1], 'bo')  # Point A
plt.plot(b[0], b[1], 'ro')  # Point B
plt.text(a[0]-0.3, a[1]+0.2, 'A (1,3)', fontsize=10, weight='bold')
plt.text(b[0]+0.1, b[1]+0.1, 'B (5,4)', fontsize=10, weight='bold')

# Limits and labels
plt.xlim(-1, 7)
plt.ylim(-1, 7)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('2D Vector Addition Sketch')
plt.legend(loc='upper left')

plt.show()
````

### Q8. Given two 2D vectors `u` and `v` where `u = [-1, 4]` and `v = [3, -2]`, find the dot product `u . v`, the angle between `u` and `v`, and the projection of `u` onto `v`.

````python
u = np.array([-1, 4])
v = np.array([3, -2])

dot_product_of_u_and_v = u @ v
print(dot_product_of_u_and_v)

cos_theta_u_v = (dot_product_of_u_and_v) / ((np.linalg.norm(u)) * (np.linalg.norm(v)))

projection_u_v = (np.dot(u, v) / np.dot(v, v)) * v

print("dot product", dot_product_of_u_and_v)
print("cos_theta_u_v", cos_theta_u_v)
print("projection_u_v", projection_u_v)
````

### Q10. Find a nearest point from a given array of points to a point P1.
P1 - (3,9,3)

Array of points :
```
[[1, 8, 5],
[8, 8, 2],
[7, 9, 1],
[5, 6, 3],
[9, 1, 2],
[6, 5, 9],
[5, 8, 6],
[3, 4, 8],
[6, 8, 4],
[7, 2, 7]]
```

````python
p1 = np.array([3, 9, 3])

array_of_points = np.array([
    [1, 8, 5],
    [8, 8, 2],
    [7, 9, 1],
    [5, 6, 3],
    [9, 1, 2],
    [6, 5, 9],
    [5, 8, 6],
    [3, 4, 8],
    [6, 8, 4],
    [7, 2, 7]    
])

# Usning brute force
nearest_point = []
nearest_point_distance = 99999
for point in array_of_points:
    current_dist = np.sqrt(np.sum(np.square(p1 - point)))
    print(current_dist)
    if current_dist < nearest_point_distance:
        nearest_point_distance = current_dist
        nearest_point = point

print("nearest point to p1", nearest_point)

# using numpy way
distances = np.linalg.norm(array_of_points - p1, axis=1)

min_index = np.argmin(distances) # get index of lowest 

print("neares point", array_of_points[min_index])
````
